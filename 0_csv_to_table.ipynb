```python
# Parâmetros de entrada
csv_path = "/Volumes/workspace/my_database/csv_data/dirty_cafe_sales.csv"
output_database = 'my_database'
output_table_name = 'dirty_cafe_sales_table'

df = spark.read.csv(
    csv_path,
    header=True,
    inferSchema=True # para inferir o tipo da coluna
    )

df.printSchema() # OBS: mesmo com o inferschema está lendo tudo como string, talvez pelos erros e vazios
df.show(5)

# Renomeia colunas para remover espaços (necessário para Delta)
for col_name in df.columns:
    df = df.withColumnRenamed(col_name, col_name.replace(' ', '_').lower())

print(f"\nNew columns names: {df.columns}")

# Salvamento
output_table = f'workspace.{output_database}.{output_table_name}'

df.write.saveAsTable(
    name= output_table,
    mode='overwrite',
    partitionBy=None
    )

print(f"\nTabela criada com sucesso no Catalog, em: {output_table}")
```
